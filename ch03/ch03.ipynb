The smallest architecture outlined in the paper from the GPT-2 team includes 124M parameters, with the following features.

```python
GPT_CONFIG_124M = {
    "vocab_size": 50257,    # Vocabulary size
    "context_length": 1024, # Context length
    "emb_dim": 768,         # Embedding dimension
    "n_heads": 12,          # Number of attention heads
    "n_layers": 12,         # Number of layers
    "drop_rate": 0.1,       # Dropout rate
    "qkv_bias": False       # Query-Key-Value bias
}
```

First, make sure you import `tiktoken`. We use `gpt2` weights for the BPE algorithm.

```python
import torch
import tiktoken

tokenizer = tiktoken.get_encoding("gpt2")

batch = []

# test text
txt1 = "Every effort moves you"
txt2 = "Every day holds a"

batch.append(torch.tensor(tokenizer.encode(txt1)))
batch.append(torch.tensor(tokenizer.encode(txt2)))
batch = torch.stack(batch, dim=0)
print(batch)
```

The following class builds a basic 124M GPT-2 architecture, although it misses many necessary functions and details. Necessary annotations are added.

```python
import torch.nn as nn

# the transformer block of this version don't do anything
class DummyTransformerBlock(nn.Module):
    def __init__(self, cfg):
        super().__init__()

    def forward(self, x):
        return x 

# the layernorm block of this version don't do anything
class DummyLayerNorm(nn.Module):
    def __init__(self, normalized_shape, eps=1e-5):
        super().__init__()

    def forward(self, x):
        return x 

class GPTModelV1(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        # 1. token embedding (token IDs -> vectors)
        self.tok_emb = nn.Embedding(cfg["vocab_size"], cfg["emb_dim"])
        # 2. position embedding (absolute position encoding)
        self.pos_emb = nn.Embedding(cfg["context_length"], cfg["emb_dim"])
        # 3. dropout
        self.dropout = nn.Dropout(cfg["drop_rate"])

        # 4. 12 Transformer blocks(Simplied)
        self.trf_blocks = nn.Sequential(*[DummyTransformerBlock(cfg) for _ in range(cfg["n_layers"])])

        # 5. LayerNorm(Simplied)
        self.layernorm = DummyLayerNorm(cfg["emb_dim"])

        # 6. Output decoder
        self.out_head = nn.Linear(cfg["emb_dim"], cfg["vocab_size"], bias=False)

    # tokenized input -> embedding -> dropout -> transformer_block -> layernorm -> output
    def forward(self, in_idx):
        batch_size, seq_len = in_idx.shape
        tok_embeds = self.tok_emb(in_idx)
        pos_embeds = self.pos_emb(torch.arange(seq_len, device=in_idx.device))
        # use broadcast mechanism to add embeddings
        x = tok_embeds + pos_embeds
        x = self.dropout(x)
        x = self.trf_blocks(x)
        x = self.layernorm(x)
        logits = self.out_head(x)
        return logits
```

```python
torch.manual_seed(325)
model = GPTModelV1(GPT_CONFIG_124M)

logits = model(batch)
print("Output shape:", logits.shape)
print(logits)
```

## 3.2 LayerNorm

**LayerNorm** centers the activations of a neural network layer around a mean of $0$ and normalizes their variance to $1$. 
- This stabilizes training and enables faster convergence to effective weights.
- LayerNorm is usually applied when
    - Before and after the multi-head attention layer
    - Before the final output layer

For a set of elements with mean $\mu$ and variance $\sigma^2$, the normalization is described as
$$x'=\gamma\cdot\frac{x-\mu}{\sigma}+\beta$$

- LayerNorm normalizes independently on **each sample** within a batch, which is different from BatchNorm and other normalization methods.
- $\gamma$ and $\beta$ are separately called `scale` and `shift`.
    - At first, they are set to $1$ and $0$
    - They are trainable parameters and they change if the model determines that doing so would improve its performance

```python
class LayerNorm(nn.Module):
    def __init__(self, emb_dim):
        super().__init__()
        self.eps = 1e-5
        # trainable parameters using nn.Parameter
        self.scale = nn.Parameter(torch.ones(emb_dim))
        self.shift = nn.Parameter(torch.zeros(emb_dim))

    def forward(self, x):
        # LayerNorm normalized by samples, that is, dim=-1
        mean = x.mean(dim=-1, keepdim=True)
        var = x.var(dim=-1, keepdim=True, unbiased=False) # unbiased variance uses n-1 in the denominator
        norm_x = (x - mean) / torch.sqrt(var + self.eps) # small eps to prevent /0 error
        return self.scale * norm_x + self.shift
```

## 3.3 Feed forward network with GELUs

In deep learning, activation functions such as **ReLU**, sigmoid were commonly used in the models years ago. In LLMs, two notable examples improved from ReLU are **GeLU** and **SwiGLU**. 
- They are more complex compared with ReLU
- They usually perform better on LLMs and other models nowadays

ReLU function is defined as
$$\mathrm{ReLU}(x)=\max(0,x)$$

GeLU combines ReLU and the standard Gaussian distribution function as
$$\mathrm{GeLU}(x)=x\cdot \Phi(x)$$

In practice, GeLU is usually approximated as the following expression shows
$$\mathrm{GeLU}\approx 0.5x\left(1+\tanh\left[\sqrt{\frac{2}{\pi}}(x+0.044715x^3)\right]\right)$$

```python
class GELU(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, x):
        # approximation expression
        return 0.5 * x * (1 + torch.tanh(torch.sqrt(torch.tensor(2.0 / torch.pi)) * (x + 0.044715 * torch.pow(x, 3))))
```

We can use `matplotlib` to see the graph of GeLU.
- If you meet any errors, try a newer version of `matplotlib`

```python
import matplotlib.pyplot as plt

gelu, relu = GELU(), nn.ReLU()

# example data
x = torch.linspace(-3, 3, 100)
y_gelu, y_relu = gelu(x), relu(x)

plt.figure(figsize=(8, 3))
for i, (y, label) in enumerate(zip([y_gelu, y_relu], ["GELU", "ReLU"]), 1):
    plt.subplot(1, 2, i)
    plt.plot(x, y)
    plt.title(f"{label} activation function")
    plt.xlabel("x")
    plt.ylabel(f"{label}(x)")
    plt.grid(True)

plt.tight_layout()
plt.show()
```

With GeLU activation function, we can implement a **feed forward network**(FFN). FFN is usually considered as two fully connected linear layers.
- You may think of it as $y=W_2(W_1x+b_1)+b_2$.
- Usually, the first linear layer extends the feature dimension to $k$ times, and the second linear layer switches it back. $k$ could be $2,4,8$ and so on.

```python
class FeedForward(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        # input -> 1st nn.Linear -> GeLU -> 2nd nn.Linear (output)
        self.layers = nn.Sequential(
            nn.Linear(cfg["emb_dim"], 4*cfg["emb_dim"]),
            GELU(),
            nn.Linear(4*cfg["emb_dim"], cfg["emb_dim"])
        )

    def forward(self, x):
        return self.layers(x)
```

Do make sure that FFN doesn't change the shape of the input.

## 3.4 Residual connection

Residual connections, also known as *shortcut connections*, were introduced by [Kaiming He et. al in the **ResNet** architecture](https://arxiv.org/abs/1512.03385). The main idea is to let a network layer learn a **residual function** $F(x)=H(x)−x$ instead of directly learning the target mapping $H(x)$.
- If the model doesn't learn anything new, then $F(x)\approx 0$, which doesn't affect the original performance of the model.
- This architecture promises that **the performance won't become worse as the model becomes more complex**.
- It also reduces the gradient vanishing problem effectively.

## 3.5 Build an untrained GPT model

The basic 124M GPT-2 model used 12 transformer-decoder layers.
- The tokenized inputs are first embedded and then dropped out.
- Then, they go across 12 decoders (TransformerBlocks).
- Finally, they go across the LayerNorm and a linear output layer.

```python
class GPTModel(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        # embedding layers
        self.tok_emb = nn.Embedding(cfg["vocab_size"], cfg["emb_dim"])
        self.pos_emb = nn.Embedding(cfg["context_length"], cfg["emb_dim"])
        # dropout layer
        self.drop_emb = nn.Dropout(cfg["drop_rate"])
        # 12 transformer blocks
        self.trf_blocks = nn.Sequential(*[TransformerBlock(cfg) for _ in range(cfg["n_layers"])])
        # final process
        self.final_norm = LayerNorm(cfg["emb_dim"])
        self.out_head = nn.Linear(cfg["emb_dim"], cfg["vocab_size"], bias=False)

    def forward(self, in_idx):
        batch_size, seq_len = in_idx.shape
        tok_embeds = self.tok_emb(in_idx)
        pos_embeds = self.pos_emb(torch.arange(seq_len, device=in_idx.device))
        x = tok_embeds + pos_embeds  # Shape [batch_size, num_tokens, emb_size]
        x = self.drop_emb(x)
        x = self.trf_blocks(x)
        x = self.final_norm(x)
        logits = self.out_head(x)
        return logits
```

```python
torch.manual_seed(325)
model = GPTModel(GPT_CONFIG_124M)

out = model(batch)
print("Input batch:\n", batch)
print("\nOutput shape:", out.shape)
print(out)
```

## 3.6 Generate new texts

The model we've implemented just give a tensor output. To turn these numbers into actual text, we need to add something new.

If you've used any LLMs especially GPT, you could see that the model gives output one word by one word. This is actually how GPT works to generate texts: **One by one from left to right**. This is what we call an **autoregressive model**.

The simplest way is to **always give the output token with the highest probability**.
- The output (known as `logits`) is usually processed by softmax to return a probability distribution output of the vocabulary tokens. 
- When the model generates the first new token according to the last given context, the generated token becomes a part of the known context.
- The model keeps doing so, predicting new tokens with tokens generated by itself.

```python
def generate_text_simple(model, idx, max_new_tokens, context_size):
    # idx is (batch, n_tokens) array of indices in the current context
    for _ in range(max_new_tokens):
        
        # use the last "context_size" tokens for predictions
        idx_cond = idx[:, -context_size:]
        with torch.no_grad():
            logits = model(idx_cond)
        
        # Focus only on the last time step
        # (batch, n_tokens, vocab_size) becomes (batch, vocab_size)
        logits = logits[:, -1, :]  
        probas = torch.softmax(logits, dim=-1)
        # Get the idx of the vocab entry with the highest probability value
        idx_next = torch.argmax(probas, dim=-1, keepdim=True)

        # Append sampled index to the running sequence
        idx = torch.cat((idx, idx_next), dim=1)  # (batch, n_tokens+1)

    return idx
```

```python
start_context = "Hello, I am"

encoded = tokenizer.encode(start_context)
print("start context:", start_context)
print("encoded:", encoded)
encoded_tensor = torch.tensor(encoded).unsqueeze(0)

model.eval() # disable dropout
out = generate_text_simple(
    model=model,
    idx=encoded_tensor, 
    max_new_tokens=6, 
    context_size=GPT_CONFIG_124M["context_length"]
)
print("Output with 6 tokens predicted:", out)
print("Output length:", len(out[0]))

decoded_text = tokenizer.decode(out.squeeze(0).tolist())
print("output text:", decoded_text)
```

## 3.7 FLOPs analysis

FLOPs, also known as *FLoating Point Operations*, measures the computational complexity of certain neural network models by counting the number of floating-point operations executed per second.
- High FLOPs usually indicates more intensive computation and energy consumption.
- To continue, necessary Python packages are required, such as `thop`.
- The following code only measures the forward calculation process.
- FLOPs is different from FLOPS (*FLoating Point Operations Per Second*). The latter measures the computational ability of certain devices.

```python
from thop import profile

# GPT-2 configs for different sizes
BASE_CONFIG = {
    "vocab_size": 50257,     # Vocabulary size
    "context_length": 1024,  # Context length
    "drop_rate": 0.0,        # Dropout rate
    "qkv_bias": True         # Query-key-value bias
}

model_configs = {
    "gpt-small (124M)": {"emb_dim": 768, "n_layers": 12, "n_heads": 12},
    "gpt-medium (355M)": {"emb_dim": 1024, "n_layers": 24, "n_heads": 16},
    "gpt-large (774M)": {"emb_dim": 1280, "n_layers": 36, "n_heads": 20},
    "gpt-xl (1558M)": {"emb_dim": 1600, "n_layers": 48, "n_heads": 25},
}

# for effeciency, it's recommended to use gpu (cuda) for calculating
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
batch_size = 2
# randomly generated inputs
input_tensor = torch.randint(0, 50257, (batch_size, 1024)).to(device)

for size in model_configs:
    BASE_CONFIG.update(model_configs[size])

    model = GPTModel(BASE_CONFIG).bfloat16()
    model.to(device)

    # MACS = multiply-accumulate operations, also a type of unit for evaluating the computation ability
    # MACS are typically counted as two FLOPS (one multiply and one accumulate)
    macs, params = profile(model, inputs=(input_tensor,), verbose=False)
    flops = 2*macs
    print(f"{size:18}: {flops:.1e} FLOPs")

    del model
    torch.cuda.empty_cache()
```

## 3.8 KV cache

Quick review: in the attention mechanism, each embedded token $x_i$ is projected into $q_i,k_i,v_i$. For a certain query $q_i=x_iW_q$, dot product with $K=XW_k$ and $V=XW_v$ is required.
- Due to the causal mask attention, for input $x_i$:
    - Query $q_i$ is necessary.
    - Keys $k_1,\cdots,k_i$ and values $v_1,\cdots,v_i$ are necessary.
- However, keys and values for $x_1,\cdots,x_{i-1}$ are computed when processing input $x_{i-1}$. This brings poor efficiency without caching previous keys and values.

So, **KV cache** is a method to store calculated keys and values for reuse. This speeds up the computation but requires extra memory for the cache. There are many ways to realize KV cache, and you can refer to `[Root]/extras/gpt_with_kv_cache.py` for improvements.

You can also see the summary below.
1. Registering the cache buffers in the `MultiHeadAttention` class.
2. Adding `using_cache` option in `forward(self, x)`.
3. Clearing the cache by resetting buffers in the `MultiHeadAttention` class.
4. Propagating `use_cache` in the full model.
